```python
from gps_denied_landing import run_simulation
result = run_simulation('flat_obstacles', seed=2000)
result['metrics'], result['selected']
```

```python
import matplotlib.pyplot as plt
arrays = result['arrays']
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
for ax, key, title in zip(axes, ['terrain', 'truth_safe', 'predicted_safe'], ['Terrain', 'Truth-safe centers', 'Predicted-safe centers']):
    ax.imshow(arrays[key], origin='lower')
    ax.set_title(title)
if result['selected']:
    axes[2].plot(result['selected']['col'], result['selected']['row'], 'r*', markersize=14)
plt.show()
```
